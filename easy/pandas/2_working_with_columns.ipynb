### `rename()` Returns a New DataFrame

`rename()` returns the renamed DataFrame as a new copy without changing the original.
**To keep that version for later steps, store it**: `df = df.rename(...)`.

### Sometimes you want to know what values exist in a column, not see every row. 
The `drop_duplicates()` method **removes duplicate rows** from your results.  
Can work with multiple columns (unnique combinations) like **distinct** in SQL.

H `.unique`δουλεύει με μία στήλη την φορά και επιστρέφει array

```python
netflix[['country']].drop_duplicates()   #select distinct country from netflix
```

```python
netflix[['plan','country']].drop_duplicates() #select distinct plan, country from netflix 
```

`drop_duplicates()` vs. `unique()`

`drop_duplicates()` works on DataFrames and returns a DataFrame. 
For a single column, you can also use `df['col'].unique()`, which returns a NumPy array of unique values. 
Use `drop_duplicates()` when you want to keep the DataFrame structure; 
use `unique()` when you just need a quick list of distinct values.

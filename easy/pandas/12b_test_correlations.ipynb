```python
#Βρες το ποσοστό (%) των φοιτητών που πέρασαν (pass) έναντι όσων κόπηκαν (fail) 
(students['result'].value_counts(normalize=True).round(3)*100).astype(str)+' %'
```

```python
((students.groupby('result').size()/len(students)).round(3)*100).astype(str)+' %'
```

```python
students.groupby('tutoring')['result'].value_counts(normalize=True) * 100
```

```python
#Ομαδοποίησε με .groupby('tutoring') και υπολόγισε τον μέσο όρο (mean), 
# τη διάμεσο (median) και το πλήθος (count) του final_score.
students.groupby('tutoring')['final_score'].agg(['mean','median','count']).round(2)
```

```python
pd.crosstab(index=students['tutoring'],
            columns=students['result'],
            values=students['result'],
            aggfunc='count',
            normalize='index').round(3)*100
```

```python
# Βρες πόσοι φοιτητές κόπηκαν (result == 'fail') παρόλο που διάβαζαν πάνω από 10 ώρες 
# (study_hours_week > 10) και είχαν παρουσίες πάνω από 80% (attendance_pct > 80).
students.loc[
    (students['result']=='fail') &
    (students['study_hours_week']>10) &
    (students['attendance_pct']>80),
    'result'
].count()
```

```python
from scipy import stats

with_tutor = students[students['tutoring'] == 1]['final_score']
no_tutor = students[students['tutoring'] == 0]['final_score']

t_stat, p_val = stats.ttest_ind(with_tutor, no_tutor, equal_var=False)
print(f"t-statistic: {t_stat:.3f}, p-value: {p_val:.4f}")
```

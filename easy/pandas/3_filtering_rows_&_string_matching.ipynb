#### Filtering in pandas works by creating a True/False mask and passing it inside brackets. Every row where the condition is True stays; everything else is dropped.

The condition creates a Boolean Series  
mask = df["department"] == "HR"

 True for HR rows, False for everything else

 Pass the mask inside brackets to filter  
 df[mask]

 Or combine in one step (most common)

`df[df["department"] == "HR"]`  
<<from df table where df's department column='HR'>>

### 📝 String Matching στα Pandas (`.str` Accessor)

Όταν ψάχνουμε μερικό ταίριασμα κειμένου (αντί για ακριβές `==`), χρησιμοποιούμε τον `.str` accessor.

#### 🔹 Βασικές Μέθοδοι (Case-Sensitive)
*   **Ξεκινάει από:** `df[df["name"].str.startswith("Jo")]`
*   **Περιέχει οπουδήποτε:** `df[df["name"].str.contains("son")]`
*   **Τελειώνει σε:** `df[df["email"].str.endswith("@gmail.com")]`

---

#### ⚠️ 2 SOS Tips για το StrataScratch (Αποφυγή Bugs)

1.  **Διαχείριση Κενών Τιμών (`NaN`):** Αν ο πίνακας έχει κενά, ο κώδικας θα κρασάρει. Χρησιμοποιούμε πάντα `na=False`.  
    *Παράδειγμα:* `df[df["email"].str.endswith("@gmail.com", na=False)]`
2.  **Αγνόηση Πεζών/Κεφαλαίων:** Αντί για `.str.lower()`, χρησιμοποιούμε την παράμετρο `case=False` για μεγαλύτερη ταχύτητα.  
    *Παράδειγμα:* `df[df["name"].str.contains("john", case=False, na=False)]`

---

#### 🔄 Αντιστοίχιση Pandas vs SQL

| Λειτουργία | Pandas (`.str`) | SQL (`WHERE`) |
| :--- | :--- | :--- |
| **Starts with** | `.str.startswith("Jo")` | `LIKE 'Jo%'` |
| **Contains** | `.str.contains("son")` | `LIKE '%son%'` |
| **Ends with** | `.str.endswith(".com")` | `LIKE '%.com'` |
| **Case-Insensitive** | `case=False` | `ILIKE` (PostgreSQL) |


```python
netflix[netflix['country'].str.contains('GR',case=False, na=False)]
#where country ILIKE '%GR%'
```

```python
# Το ^N σημαίνει: "να ξεκινάει από N"
netflix[netflix['country'].str.contains('^N', case=False, na=False)]
```

```python
# Το s$ σημαίνει: "να τελειώνει σε s"
netflix[netflix['country'].str.contains('s$', case=False, na=False)]
```

```python
netflix.loc[netflix['country'].str.contains('s$'),['country','hours_watched']]
#.str.endswith('s')
```

```python
netflix[netflix['country'].str.lower()=='germany']
```

```python
netflix[netflix['country'].str.lower().str.contains('^n')]
```

# Σημειώσεις Pandas: Υπό Συνθήκη Καταμέτρηση (Conditional Counting)

Όταν θέλουμε να μετρήσουμε πόσες εγγραφές (rows) ικανοποιούν συγκεκριμένα κριτήρια σε ένα DataFrame, έχουμε δύο βασικά εργαλεία: τα **Boolean Masks** και το **Groupby**. 

---

## 1. Η Μέθοδος: Boolean Mask + `.sum()` 
*Σαν το SUM(CASE WHEN ...) της SQL*

Αυτή η τεχνική βασίζεται στο γεγονός ότι στην Python (και στο Pandas), οι λογικές τιμές συμπεριφέρονται ως αριθμοί:
* `True` = `1`
* `False` = `0`

Όταν εφαρμόζουμε μια συνθήκη σε μια στήλη, το Pandas δημιουργεί μια σειρά από `True` και `False`. Αν προσθέσουμε αυτή τη σειρά με την `.sum()`, το αποτέλεσμα είναι το πλήθος των `True` (δηλαδή των γραμμών που ικανοποιούν τη συνθήκη).

### Παράδειγμα Κώδικα:
```python
import pandas as pd

# Δημιουργία δοκιμαστικού DataFrame
data = {
    "name": ["Alice", "Bob", "Charlie", "David", "Eva"],
    "department": ["HR", "IT", "HR", "IT", "Sales"],
    "salary": [45000, 60000, 52000, 48000, 70000]
}
df = pd.DataFrame(data)

# 1. Καταμέτρηση με Boolean Mask
hr_count = (df["department"] == "HR").sum()
high_salary_count = (df["salary"] > 50000).sum()
total = len(df)

# 2. Συγκέντρωση αποτελεσμάτων σε ένα νέο DataFrame (Οριζόντια δομή)
summary_df = pd.DataFrame({
    "Total Employees": [total],
    "HR Count": [hr_count],
    "High Earners (>50k)": [high_salary_count]
})

print(summary_df)
```

---

## 2. Η Μέθοδος: `groupby()` + `.size()`

Αν ο στόχος μας είναι να σπάσουμε τα δεδομένα με βάση τις **υπάρχουσες κατηγορίες** μιας στήλης (π.χ. να δούμε το πλήθος για *κάθε* τμήμα ξεχωριστά), η χρήση του `groupby` είναι η πιο γρήγορη και αποδοτική λύση.

### Παράδειγμα Κώδικα:
```python
# Καταμέτρηση ανά τμήμα αυτόματα (Κάθετη δομή)
dept_counts = df.groupby("department").size()

print(dept_counts)
```

---

## ⚔️ Σύγκριση: Πότε επιλέγω τι;

| Κριτήριο | Boolean Mask (`.sum()`) | Groupby (`.groupby()`) |
| :--- | :--- | :--- |
| **Στόχος** | Μέτρηση **διαφορετικών ή custom συνθηκών** (π.χ. και για τμήμα και για μισθό μαζί). | Ομαδοποίηση με βάση **τις τιμές μιας συγκεκριμένης στήλης** (π.χ. ανά τμήμα). |
| **Δομή Αποτελέσματος** | **Οριζόντια (Columns):** Φτιάχνεις custom στήλες για κάθε μέτρηση. | **Κάθετη (Rows):** Κάθε κατηγορία γίνεται μια ξεχωριστή σειρά στο αποτέλεσμα. |
| **Ευελιξία** | Πολύ υψηλή. Μπορείς να συνδυάσεις άσχετα κριτήρια μεταξύ τους. | Περιορισμένη στις κατηγορίες που ήδη υπάρχουν στα δεδομένα σου. |

### Χρυσός Κανόνας:
* Αν θέλεις report τύπου: *«Πόσους έχουμε στο IT, πόσους στο HR, πόσους στο Sales;»* ➡️ **`groupby`**
* Αν θέλεις report τύπου: *«Πόσοι είναι στο HR **ΚΑΙ** πόσοι (γενικά) βγάζουν πάνω από 50k;»* ➡️ **`Boolean Mask`**

---

## 💡 Advanced Tip: Ποσοστά αντί για Μόνες Μετρήσεις
Αν αντικαταστήσεις την `.sum()` με την **`.mean()`** (μέσος όρος) σε ένα Boolean Mask, το Pandas θα σου επιστρέψει το **ποσοστό (%)** των γραμμών που ικανοποιούν τη συνθήκη!

```python
# Επιστρέφει το ποσοστό των υπαλλήλων στο HR (π.χ. 0.40 για 40%)
hr_percentage = (df["department"] == "HR").mean()
```

````python
# Θέλουμε να μάθουμε πόσες παραγγελίες έγιναν συνολικά μέσω της εφαρμογής (app)
(retail['channel']=='app').sum()
#retail[retail['channel']=='app'].groupby('channel')['channel'].count().reset_index(name='count')
````

````python
# Ζητούμενο: Θέλουμε να δούμε την κινητικότητα στις περιοχές East και North.
# Η πρόκληση:
# 	1. Φτιάξε ένα custom DataFrame που θα δείχνει οριζόντια (σε στήλες) πόσες παραγγελίες 
#       έγιναν στην περιοχή East και πόσες στην περιοχή North χρησιμοποιώντας Boolean Masks.

dic={'East':[(retail['region']=='East').sum()],
      'North':[(retail['region']=='North').sum()]
      }
east_north_count = pd.DataFrame(data=dic, index=['count'])
east_north_count
````

````python
# 	2. Γράψε μια δεύτερη, εναλλακτική γραμμή κώδικα χρησιμοποιώντας groupby() που 
#       επιστρέφει το πλήθος των παραγγελιών για όλες τις περιοχές (region) κάθετα

retail.groupby('region')['region'].count().reset_index(name='value_count')
````

````python
# Ζητούμενο: Θέλουμε να αναλύσουμε τις μεγάλες παραγγελίες στην κατηγορία των ηλεκτρονικών (Electronics).
# Η πρόκληση:
# 	1. Βρες το ποσοστό (%) των συνολικών παραγγελιών που ανήκουν στην κατηγορία Electronics. (Θυμήσου το Advanced Tip με την .mean()).

result = (retail['category']=='Electronics').mean()*100 
print(result,'%')
````

````python
electronics_count = (retail['category']=='Electronics').sum()
total_count = len(retail)  # ή retail['category'].size  #αποφεύγω την .count γιατι δεν μετράει τις null τιμές αν υπάρχουν

dict={'electronics': electronics_count,
      'total': total_count,
      'electronics_pct':[100*electronics_count/total_count]
      }
df = pd.DataFrame(data=dict)
df
````

````python
# 	2. Μέτρα πόσες παραγγελίες έγιναν στην κατηγορία Home 
#       ΚΑΙ είχαν έσοδα (revenue) μεγαλύτερα από 1000

((retail['category']=='Home') & (retail['revenue']>1000)).sum()
#retail[(retail['category']=='Home') & (retail['revenue']>1000)]['category'].count()
````

````python
# Ζητούμενο: Θέλουμε μια πλήρη οικονομική εικόνα ανά κατηγορία προϊόντος (category).

# Η πρόκληση: Χρησιμοποίησε τη groupby() στην κατηγορία category, αλλά αυτή τη φορά συνδύασέ 
# την με την .agg() για να βρεις για κάθε κατηγορία:
#    Το συνολικό πλήθος των παραγγελιών (size)
#    Το άθροισμα των συνολικών εσόδων (revenue)
#    Τον μέσο όρο των τεμαχίων που αγοράστηκαν (units)

retail.groupby('category').aggregate(values_count=('category','count'),
                                      sum_rev=('revenue','sum'),
                                      avg_units=('units','mean')).round(1).reset_index()                            

````

````python
#Βρες το ποσοστό (%) των συνολικών παραγγελιών ανα κατηγορία

pd.crosstab(retail['category'],columns='pct', normalize=True).round(2)*100
````

````python
#Βρες το ποσοστό (%) των συνολικών παραγγελιών ανα κατηγορία
((retail.groupby('category').size()/len(retail)).round(2)*100).reset_index(name='pct')
````

````python
#Βρες το ποσοστό (%) των συνολικών παραγγελιών ανα κατηγορία
(retail['category'].value_counts(normalize=True).round(2)*100).reset_index(name='pct')
````

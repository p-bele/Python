### Missing values can cause bugs and unexpected results  

`NaN` stands for "Not a Number", but it really means **"missing" or "unknown"**.

**NaN vs None**  
In pandas, **both** NaN (from NumPy) and None (Python’s built-in) **represent missing data**. Pandas treats them interchangeably in most situations. You’ll see both, and the same methods handle both.

---

Finding Missing Values: `.isna()`  
`df[df['phone_number'].isna()]`

To find rows where a value exists, use `.notna()` — **the opposite of .isna()**

**Counting Missing Values**
Before cleaning data, it helps to know how much is missing. A few useful patterns:

Count NaN values per column  
`df.isna().sum()`

ratio missing per column  
`df.isna().mean()` 

`.isna().sum()` counts missing values in each column. `.isna().mean()` gives you the proportion. These are the first things to run when exploring a new dataset.

```python
import pandas as pd
import numpy as np

# Δημιουργία δεδομένων με κενά (np.nan)
data = {
    'name': ['Anna', 'John', 'Peter', 'Maria', 'Nick'],
    'age': [25, np.nan, 19, 31, np.nan],          # Έχει NaN στη θέση 1 και 4
    'department': ['IT', 'HR', 'IT', 'Marketing', 'HR'],
    'salary': [95000, 48000, np.nan, 72000, 52000] # Έχει NaN στη θέση 2
}

# Μετατροπή σε DataFrame
df = pd.DataFrame(data)

# Εμφάνιση του πίνακα
df
```

```python
df.loc[df['age'].notna(), ['name','age']]
```

```python
df[df['salary'].isna()]
```

```python
df.isna().sum()  #μετράει τα nan values ανά στήλη
```

```python
df.isna().mean()*100  #percentage of null values
```

Το εξωτερικό (διπλό) df[...] χρησιμοποιείται μόνο όταν θέλεις να κάνεις φιλτράρισμα (boolean indexing), δηλαδή όταν θέλεις να κρατήσεις και να εμφανίσεις ολόκληρες τις γραμμές του πίνακα που πληρούν μια προϋπόθεση.

```python
df[['age']].isna().sum()
```

 μπορείς να γεμίσεις τις κενές τιμές (NaN) σε ένα DataFrame χρησιμοποιώντας τη συνάρτηση 
 `.fillna().`

```python
df['age']=df['age'].fillna('Unknown')
df[['age']]
```

```python
#γεμίζω την κενή τιμή του salary με την διαμεσο της στηλης
df['salary']=df['salary'].fillna(df['salary'].median())
df[['salary']]
```

Dropping Missing Values with `.dropna()`

Drop rows where ANY column is NaN  
`df.dropna()`

Drop rows where a SPECIFIC column is NaN  
`df.dropna(subset=["phone_number"])`

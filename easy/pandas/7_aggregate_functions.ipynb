`len(df)` counts total rows  
`df.count()` counts non null values per column  
`df['phone_number'].count()` counts non null values in specific column  
`df['department'].nunique()` counts unique values in a column  

`len(df)` counts all rows including NaN. `.count()` counts non-null values. `.nunique()` counts unique non-null values.

```python
import pandas as pd
```

```python
movies= pd.read_csv('../data/movies.csv', index_col=0)
```

```python
movies.head(3)
```

```python
len(movies) #total rows
```

```python
movies.count() #non-null values per column
```

```python
movies['genre'].count() #non-null values in column 'genre' 
```

```python
movies['genre'].nunique() #counts the number of unique values in 'genre' column
```

```python
#μέτρα τις non null γραμμές της στήλης genre, όπου το genre είναι Thriller 
movies.loc[movies['genre']=='Thriller','genre'].count()
```

```python
#μέτρα τις μοναδικές τιμές χρονολογιών όπου το 'genre' είναι Thriller
movies.loc[movies['genre']=='Thriller','year'].nunique()
```

`df['salary'].sum()` αθροίζει τις τιμές της στήλης salary 

```python
movies.loc[movies['genre']=='Thriller', 'budget_musd'].sum()  #millions
```

`min()`  
`max()`  
`mean()`

```python
movies.loc[
    (movies['genre']=='Thriller') &
    (movies['rating']>7),
    'revenue_musd'
].mean().round(2)
```

```python
movies.loc[movies['genre']=='Action', 'runtime_min'].max()  #max διάρκεια ταινίας
```

```python
movies.loc[movies['genre']=='Action', 'runtime_min'].describe().round(2)
```

#### Multiple Aggregates at Once with .agg()  
Instead of calling each method separately, .agg() lets you run multiple aggregations in one call: `df['salary'].aggregate(['max','min','mean','median'])`

```python
movies.loc[
    movies['genre']=='Action',
    'revenue_musd'
].aggregate(['mean','median','std','skew']).round(2)
```

```python
movies['revenue_musd'].quantile([0.25, 0.5, 0.75])
```

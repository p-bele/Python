#### κάτι που πρέπει να θυμάσαι! 

• Η **Pandas** εκτελείται διαδοχικά (Procedural/Imperative): Ακολουθεί πιστά τη σειρά που γράφεις τις εντολές από αριστερά προς τα δεξιά (σαν γραμμή παραγωγής). 

• Η **SQL** εκτελείται δηλωτικά (Declarative): Εσύ της λες τι θες να σου φέρει, αλλά η ίδια η βάση (ο Query Optimizer) ανακατεύει τη σειρά των λέξεων (SELECT, FROM, WHERE) για να κάνει τον υπολογισμό με τον πιο γρήγορο τρόπο.


#### sorting by string length

Strings can’t be sorted by length directly — first create a column to sort on.  
`.str.len()` returns the length of each string in a column

df["name_len"] = df["first_name"].str.len()  
df.sort_values(by=['name_len'],ascending=False)

### shortcut for sorting with top N simultaneously  
`.nlargest(n, col)` / `.nsmallest(n, col)`  
These are equivalent:  
df.sort_values("salary", ascending=False).head(5)  
df.nlargest(5, "salary")

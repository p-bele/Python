```python
cars = cars.rename(str.lower,axis='columns')
```

```python
cars.head(2)
```

```python
cars['price']= cars['price'].str.replace('Rs','').astype(float)
```

```python
cars['price'].describe().round(2)
```

```python
cars = cars.dropna(subset=['price'])  # σβήνω γραμμή αν δεν υπάρχει τιμή
```

```python
cars= cars.dropna(subset=['make'])   # σβήνω γραμμή αν δεν υπάρχει μάρκα
```

```python
cars['colour']= cars['colour'].fillna('Missing') #αν δεν υπάρχει colour βάζω Missing
```

```python
cars['doors']=cars['doors'].fillna('Missinig')  #αν δεν υπάρχει doors βάζω Missing
```

```python
cars['odometer (km)']= cars['odometer (km)'].fillna(cars['odometer (km)'].mean()) 
```

```python
cars.info()
```

```python
pd.crosstab(cars['make'],cars['colour'], margins=True)
```

```python
students['score_label']= np.where(students['final_score']<50, 'below_base', 'above_base')
students.head(5)
```

```python
conditions = [students['sleep_hours']<4,
              students['sleep_hours']<5,
              students['sleep_hours']<6,
              students['sleep_hours']<7,
              students['sleep_hours']<8]

choices = ['very poor sleep',
           'poor sleep',
           'fair sleep',
           'good',
           'very good']

students['sleep_label'] = np.select(conditions, choices, default='excellent')
students
```

```python
df = pd.crosstab(students['sleep_label'], students['result'], normalize='index').round(2)*100
df.sort_values(by='pass', ascending=False)
```

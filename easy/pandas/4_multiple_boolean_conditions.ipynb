### 🔀 Συνδυασμός Συνθηκών στα Pandas (Combining Conditions)

Όταν θέλουμε να φιλτράρουμε ένα DataFrame με βάση πολλαπλά κριτήρια, χρησιμοποιούμε τους λογικούς τελεστές των Pandas και τη μέθοδο `.isin()`.

#### ⚠️ 2 Χρυσοί Κανόνες 
1. **Παρενθέσεις `()` παντού:** Κάθε ξεχωριστή συνθήκη πρέπει υποχρεωτικά να μπαίνει μέσα σε παρενθέσεις, αλλιώς η Python θα πετάξει σφάλμα (`ValueError`).
2. **Όχι `and`, `or`, `not`:** Στα Pandas χρησιμοποιούμε αποκλειστικά τα σύμβολα `&`, `|`, `~` (bitwise operators) γιατί ελέγχουν ολόκληρη τη στήλη στοιχείο προς στοιχείο (element-wise).

---

### 🔹 1. Ο τελεστής ΚΑΙ (`&` - AND)
Επιστρέφει τις γραμμές που ικανοποιούν **όλες** τις συνθήκες ταυτόχρονα.
```python
# Υπάλληλοι στο τμήμα 'IT' ΚΑΙ με μισθό πάνω από 100.000
df[(df['department'] == 'IT') & (df['salary'] > 100000)]
```

### 🔹 2. Ο τελεστής Η (`|` - OR)
Επιστρέφει τις γραμμές που ικανοποιούν **τουλάχιστον μία** από τις συνθήκες.
```python
# Υπάλληλοι που είναι στο τμήμα 'HR' Η στο τμήμα 'Marketing'
df[(df['department'] == 'HR') | (df['department'] == 'Marketing')]
```

### 🔹 3. Ο τελεστής ΑΡΝΗΣΗΣ (`~` - NOT)
Αντιστρέφει τη συνθήκη. Επιστρέφει ό,τι **ΔΕΝ** ταιριάζει με το κριτήριο (χρήσιμο και με string methods).
```python
# Υπάλληλοι που το όνομά τους ΔΕΝ περιέχει το γράμμα 'o'
df[~df['first_name'].str.contains('o')]
```

---

### 🎯 4. Η μέθοδος `.isin()` (Αντικατάσταση των πολλών OR)
Όταν θέλουμε να ελέγξουμε αν μια τιμή ανήκει σε μια λίστα επιλογών, αντί να γράφουμε πολλά `|` (OR), χρησιμοποιούμε την `.isin()`. Είναι πιο καθαρή, πιο γρήγορη και δέχεται Python λίστα `[]`.

```python
# Αντί για: (df['city'] == 'Athens') | (df['city'] == 'Paris') | (df['city'] == 'London')
# Γράφουμε το κομψό:
my_cities = ['Athens', 'Paris', 'London']
df[df['city'].isin(my_cities)]
```

#### 💡 Συνδυασμός `~` και `.isin()` (NOT IN):
Αν θέλουμε να βρούμε όσους **ΔΕΝ** ανήκουν στη λίστα, βάζουμε το `~` στην αρχή της συνθήκης:
```python
# Υπάλληλοι που ΔΕΝ μένουν σε αυτές τις πόλεις
df[~df['city'].isin(my_cities)]
```

---

### 🔄 Αντιστοίχιση Pandas vs SQL

| Λειτουργία | Pandas | SQL |
| :--- | :--- | :--- |
| **ΚΑΙ** | `&` | `AND` |
| **Η** | `\|` | `OR` |
| **ΑΡΝΗΣΗ** | `~` | `NOT` |
| **ΛΙΣΤΑ ΕΠΙΛΟΓΩΝ** | `.isin([...])` | `IN (..._` |
| **ΕΚΤΟΣ ΛΙΣΤΑΣ** | `~df[...].isin([...])` | `NOT IN (...)` |

### μπορείς να συγκρίνεις δύο στήλες μεταξύ τους, γραμμή-γραμμή τόσο στην pandas όσο και στην SQL!!!

````python
#σύγκριση δύο στηλών μεταξύ τους

movies[movies['budget_musd']<movies['revenue_musd']]

# SELECT * FROM movies WHERE budget_musd < revenue_musd
````

#### ΠΡΟΣΟΧΗ ΣΤΙΣ ΠΑΡΕΝΘΕΣΕΙΣ!

````python
movies[
    ((movies['budget_musd']<movies['revenue_musd']) &
    (movies['votes']>50000)) |
    (movies['runtime_min']<90)
    ]

# SELECT *
# FROM movies
# WHERE (revenue_musd > budget_musd AND votes > 50000)
#    OR runtime_min < 90;
````

#### `.between(value1, value2)`
both sides inclusive

````python
movies[movies['runtime_min'].between(90,120)]
````

````python
# μπορώ να αποθηκεύσω την μάσκα σε ξεχωριστές μεταβλητές για πολύπλοκα ερωτήματα
year_2017= movies['year']==2017
is_drama= movies['genre']=='Drama'

movies[(year_2017 & is_drama)]
````

````python
netflix= pd.read_csv("../data/netflix_streams.csv",index_col=0)
````
